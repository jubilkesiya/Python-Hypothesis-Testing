### ESTIMATION AND CONFIDENCE INTERVALS

Data given: 
A total of 15 print-heads were randomly selected and tested until failure. The durability of each print-head (in millions of characters) was recorded as follows:
1.13, 1.55, 1.43, 0.92, 1.25, 1.36, 1.32, 0.85, 1.07, 1.48, 1.20, 1.33, 1.18, 1.22, 1.29


##### Assignment Tasks

a. Build 99% Confidence Interval Using Sample Standard Deviation

```python
import numpy as np
import scipy.stats as stats
import statsmodels
import warnings
warnings.filterwarnings('ignore')
```

```python
#given data
n=15 #sample size

readings=[1.13, 1.55, 1.43, 0.92, 1.25, 1.36, 1.32, 0.85, 1.07, 1.48, 1.20, 1.33, 1.18, 1.22, 1.29] 

x_bar=sum(readings)/n # sample mean=total/number
sam_std=np.std(readings, ddof=1) # sample standard deviation
dof=14 # degree of freedom=n-1
```

```python
print(readings)
print('sample mean=',x_bar)
print('sample standard deviation=',sam_std)
```

we have n<30 and we do not have the population standard deviation. So we find the t critical value to find the confidence interval.

CI=99%   =0.99
alpha=1%=0.01
alpha/2=0.005

```python
t_critical=stats.t.ppf(0.995,14) # taking the upper tail
t_critical
```

```python
margin_of_error=t_critical*(sam_std/np.sqrt(n))
margin_of_error
```

```python
con_int_lower=x_bar-margin_of_error
con_int_upper=x_bar+margin_of_error
print('lower limit: ',con_int_lower,'upper limit:',con_int_upper)
```

```python
print(f"99% Confidence Interval: ({con_int_lower:.4f}, {con_int_upper:.4f}) million characters")
```

b. Build 99% Confidence Interval Using Known Population Standard Deviation

Since the population standard deviation (sigma) is given now, we will use z test in place of t test. This is done even though the sample size is small because we now have sigma.

```python
pop_std=0.2
```

```python
z_critical=stats.norm.ppf(0.995)
z_critical
```

```python
margin_of_error_z=z_critical*(pop_std/np.sqrt(n))
margin_of_error_z
```

```python
con_int_lower_z=x_bar-margin_of_error_z
con_int_upper_z=x_bar+margin_of_error_z
print('lower limit: ',con_int_lower_z,'upper limit:',con_int_upper_z)
```

```python
print(f"99% Confidence Interval: ({con_int_lower:.4f}, {con_int_upper:.4f}) million characters")
```
